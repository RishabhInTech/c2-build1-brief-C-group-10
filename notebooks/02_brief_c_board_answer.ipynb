## CONFIG

```python
from pathlib import Path
ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
CONFIG = {
    "brief": "Brief C", "group": 10,
    "data_pack": ROOT / "data" / "Build1_DataPack.zip",
    "clean": ROOT / "data" / "clean" / "brief_c",
    "output": ROOT / "outputs" / "brief_c",
    "action": ROOT / "data" / "clean" / "brief_c" / "actions",
    "year": 2026, "observation_cutoff": "2026-07-15",
}
print(f"Kalpa Health · {CONFIG['brief']} · Group {CONFIG['group']}")
```

```python
import json
import subprocess
import sys
from decimal import Decimal

subprocess.run([sys.executable, str(ROOT / "notebooks" / "_scripts" / "clean_brief_c.py"), "--source-dir", str(CONFIG["data_pack"]), "--output-dir", str(CONFIG["clean"])], cwd=ROOT, check=True)
subprocess.run([sys.executable, str(ROOT / "notebooks" / "_scripts" / "analyze_brief_c_reconciliation.py"), "--data-dir", str(CONFIG["clean"]), "--output-dir", str(CONFIG["output"]), "--action-dir", str(CONFIG["action"]), "--year", str(CONFIG["year"]), "--observation-cutoff", CONFIG["observation_cutoff"]], cwd=ROOT, check=True)
summary = json.loads((CONFIG["output"] / "finance_bridge_summary.json").read_text())
cohort, attempts = summary["invoice_cohort"], summary["payment_attempts_created_in_h1"]
at_close_count = cohort["invoices_with_no_capture_by_h1_end"] + cohort["partially_paid_invoices_by_h1_end"]
crore = lambda value: f"₹{Decimal(value) / Decimal(10_000_000):.3f} crore"
HEADLINE = {
    "number": Decimal(cohort["positive_invoice_balance_rs_by_h1_end"]) / Decimal(10_000_000),
    "numerator": at_close_count, "denominator": cohort["invoice_count"],
}
open_pct = Decimal(cohort["remaining_positive_invoice_balance_rs_by_observation_cutoff"]) / Decimal(cohort["invoice_amount_rs"]) * 100
HEADLINE["claim"] = f"At H1 2026 close, {at_close_count:,} of {cohort['invoice_count']:,} invoices ({crore(cohort['invoice_amount_rs'])} billed) carried {crore(cohort['positive_invoice_balance_rs_by_h1_end'])} in positive balances; of {cohort['invoices_with_no_capture_by_h1_end']:,} with no captured match, {cohort['no_capture_invoices_with_capture_by_observation_cutoff']:,} received later captures ({crore(cohort['face_amount_rs_of_those_later_paid_invoices'])} face value), leaving {crore(cohort['remaining_positive_invoice_balance_rs_by_observation_cutoff'])} ({open_pct:.2f}% of H1 billings) open across {cohort['remaining_positive_balance_invoice_count_by_observation_cutoff']:,} invoices by 15 July, while {attempts['unmatched_capture_rows']:,} H1 captured attempts worth {crore(attempts['unmatched_capture_amount_rs'])} remained unmatched—Finance should prioritize June's open invoices and resolve both exception lists before calling balances lost."
print(HEADLINE["claim"])
print(f"H1 positive balance: {HEADLINE['number']:.3f} crore ({cohort['positive_invoice_balance_rs_by_h1_end']} rupees)")
print(f"Open balance as a share of H1 billing: {open_pct:.2f}%")
```

## Where Finance should start

Monthly and city/channel balances below are at 30 June. The unmatched-payment table covers H1 attempts. Keep these separate from the 15 July remaining-balance snapshot above.

```python
month = pd.read_csv(CONFIG["output"] / "finance_outstanding_by_invoice_month.csv")
city = pd.read_csv(CONFIG["output"] / "finance_outstanding_by_city_channel.csv")
unmatched = pd.read_csv(CONFIG["output"] / "finance_unmatched_capture_breakdown.csv")
june = month.loc[month["invoice_month"] == f"{CONFIG['year']}-06"].iloc[0]
close_total = month["positive_balance_at_close_rs"].sum()
june_share = june["positive_balance_at_close_rs"] / close_total * 100
print(f"June at H1 close: {int(june['open_invoices_at_close']):,} invoices, ₹{june['positive_balance_at_close_rs'] / 10_000_000:.3f} crore ({june_share:.1f}% of the close balance)")

unmatched["amount_rs"] = pd.to_numeric(unmatched["amount_rs"])
methods = unmatched.groupby("method", as_index=False).agg(attempts=("attempts", "sum"), amount_rs=("amount_rs", "sum"))
methods["amount_lakh"] = methods["amount_rs"] / 100_000
print(f"Unmatched H1 captures: {methods['attempts'].sum():,} attempts, ₹{methods['amount_rs'].sum() / 100_000:.2f} lakh")
print("\nUnmatched H1 captures by method (₹ lakh):")
print(methods[["method", "attempts", "amount_lakh"]].to_string(index=False, formatters={"amount_lakh": lambda value: f"{value:.2f}"}))
upi = methods.loc[methods["method"] == "upi", "amount_rs"].iloc[0]
print(f"UPI share of unmatched value: {upi / methods['amount_rs'].sum() * 100:.0f}%")
shapes = unmatched.groupby("reference_shape", as_index=False).agg(attempts=("attempts", "sum"), amount_rs=("amount_rs", "sum"))
print("\nUnmatched reference shapes:")
print(shapes.to_string(index=False))

city["positive_balance_at_close_rs"] = pd.to_numeric(city["positive_balance_at_close_rs"])
city_totals = city.groupby("city", as_index=False)["positive_balance_at_close_rs"].sum()
city_totals["balance_lakh"] = city_totals["positive_balance_at_close_rs"] / 100_000
print("\nPositive balances by city at H1 close (₹ lakh):")
print(city_totals.sort_values("balance_lakh", ascending=False)[["city", "balance_lakh"]].to_string(index=False, formatters={"balance_lakh": lambda value: f"{value:.2f}"}))
```

## Board-question answers

- **Why not call open invoices lost?** Captures can arrive after close, references can be malformed, and later reversals/source gaps have not been ruled out.
- **Why keep overcaptures separate?** The 704 overcaptured invoices (₹0.117 crore) do not establish that credits may be applied to another customer's invoice.
- **Is the settlement gap a bank shortfall?** No. H1 settlements net to ₹24.537 crore, and linked captures span payment dates; independent bank statements and cash registers were not supplied.
- **Where should Finance start?** At H1 close, June had 6,256 open invoices totaling ₹1.053 crore (54.1% of the positive balance); by 15 July, ₹0.755 crore remained open. Review June first, then the two local action lists. This is a priority, not a causal finding.
